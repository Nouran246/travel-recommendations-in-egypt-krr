```python
# Rules 
rules = [
    {
        "name":       "budget_match",
        "conditions": ["budget"],
        "conclusion": "affordable",
        "description": "IF user budget >= destination budget THEN affordable"
    },
    {
        "name":       "season_match",
        "conditions": ["season"],
        "conclusion": "good_timing",
        "description": "IF travel month is in destination best months THEN good_timing"
    },
    {
        "name":       "heat_match",
        "conditions": ["heat"],
        "conclusion": "heat_tolerable",
        "description": "IF user heat tolerance >= destination heat level THEN heat_tolerable"
    },
    {
        "name":       "beach_match",
        "conditions": ["beach"],
        "conclusion": "beach_ok",
        "description": "IF user beach preference matches destination beach availability THEN beach_ok"
    },
    {
        "name":       "triptype_match",
        "conditions": ["trip_type"],
        "conclusion": "right_trip_type",
        "description": "IF preferred trip type is offered by destination THEN right_trip_type"
    },
    {
        "name":       "recommend_destination",
        "conditions": ["affordable", "good_timing", "heat_tolerable", "beach_ok", "right_trip_type"],
        "conclusion": "recommend",
        "description": "IF affordable AND good_timing AND heat_tolerable AND beach_ok AND right_trip_type THEN recommend"
    },
]
```

```python
# helpers
heat_order   = {"low": 1, "moderate": 2, "high": 3, "extreme": 4}
budget_order = {"low": 1, "medium": 2, "high": 3}

month_names = {
    "january":1, "february":2, "march":3, "april":4,
    "may":5, "june":6, "july":7, "august":8,
    "september":9, "october":10, "november":11, "december":12
}

#  retrieves facts from knowledge base
def get_fact(pred, arg1):
    for f in facts:
        if f[0] == pred and f[1] == arg1:
            return f[2]
    return None

# applies packing rules to a destination
def get_packing_list(dest):
    items = []
    climate = get_fact("climate", dest)
    if climate:
        items += packing_rules.get(climate, [])
    for t in trip_types.get(dest, []):
        items += packing_rules.get(t, [])
    return list(set(items))
def check_budget(dest, prefs):
    if prefs["budget"] == "no preference":
        return True
    dest_budget = get_fact("budget_level", dest)
    return budget_order.get(prefs["budget"], 0) >= budget_order.get(dest_budget, 99)

def check_season(dest, prefs):
    if prefs["month_num"] is None or prefs["month_num"] == "no preference":
        return True
    return prefs["month_num"] in best_months.get(dest, [])

def check_heat(dest, prefs):
    if prefs["heat_tolerance"] == "no preference":
        return True
    dest_heat = get_fact("heat_level", dest)
    return heat_order.get(prefs["heat_tolerance"], 0) >= heat_order.get(dest_heat, 99)

def check_beach(dest, prefs):
    if prefs["wants_beach"] == "no preference":
        return True
    has = any(f[1] == dest for f in facts if f[0] == "has_beach")
    return has if prefs["wants_beach"] == "yes" else not has

def check_trip_type(dest, prefs):
    if prefs["trip_type"] == "no preference":
        return True
    return prefs["trip_type"] in trip_types.get(dest, [])
```

```python
# backward chaining - reads from rules list

def backward_chain(dest, prefs):
    tree_log = []
    tree_log.append(f"GOAL: recommend({dest})")

    # find the top-level rule whose conclusion is 'recommend'
    top_rule = next(r for r in rules if r["conclusion"] == "recommend")
    tree_log.append(f"  trying rule [{top_rule['name']}]: {top_rule['description']}")

    derived = {}   # stores what we've proved so far: conclusion -> True/False

    # for each condition the top rule needs, find and fire the sub-rule
    for cond in top_rule["conditions"]:
        sub_rule = next((r for r in rules if r["conclusion"] == cond), None)

        if sub_rule is None:
            # no rule for this condition, can't prove it
            derived[cond] = False
            tree_log.append(f"    [{cond}] no rule found -> FAIL")
            continue

        # the sub-rule's conditions are primitive checks against the KB
        sub_results = {}
        for prim in sub_rule["conditions"]:
            checker = condition_checkers.get(prim)
            if checker:
                result = checker(dest, prefs)
                sub_results[prim] = result
            else:
                sub_results[prim] = False

        sub_passed = all(sub_results.values())
        derived[cond] = sub_passed

        mark = "V" if sub_passed else "X"
        tree_log.append(f"    [{mark}] rule [{sub_rule['name']}] -> {cond}")
        for prim, res in sub_results.items():
            pmark = "V" if res else "X"
            tree_log.append(f"          [{pmark}] {prim} (checked against KB)")

    # fire the top rule: all sub-conclusions must be proved
    all_passed = all(derived.values())
    if all_passed:
        tree_log.append(f"  => rule [{top_rule['name']}] FIRED -> recommend({dest}) = TRUE")
    else:
        failed = [k for k, v in derived.items() if not v]
        tree_log.append(f"  => rule [{top_rule['name']}] NOT fired, failed: {failed}")

    # also return the flat checks dict for easy use by mode functions
    flat_checks = {
        "budget":    derived.get("affordable", False),
        "season":    derived.get("good_timing", False),
        "heat":      derived.get("heat_tolerable", False),
        "beach":     derived.get("beach_ok", False),
        "trip_type": derived.get("right_trip_type", False),
    }

    return all_passed, flat_checks, tree_log


def build_tree_log(prefs, city_list):
    """runs backward_chain on each city and joins the tree logs"""
    lines = []
    for dest in city_list:
        if dest not in destinations:
            continue
        _, _, tree = backward_chain(dest, prefs)
        lines += tree
        lines.append("")
    return lines
```

```python
# Fuzzy matching - finds closest matches when no perfect match exists
def get_closest_matches(results_with_fails, prefs, max_fails=2):
    """
    When no perfect matches, return cities with fewest failed rules
    max_fails: maximum number of failed rules allowed (default 2)
    """
    # Calculate fail count for each destination that had at least one check
    scored = []
    for dest in destinations:
        # Get all checks for this destination
        passed, checks, _ = backward_chain(dest, prefs)
        fail_count = sum(1 for v in checks.values() if not v)
        scored.append({
            "city": dest,
            "fail_count": fail_count,
            "passed": passed,
            "checks": checks
        })
    
    # Filter to those within max_fails
    closest = [s for s in scored if s["fail_count"] <= max_fails]
    
    # Sort by fail_count (lowest first)
    closest.sort(key=lambda x: x["fail_count"])
    
    return closest

# maps condition keys to their checker functions
condition_checkers = {
    "budget":    check_budget,
    "season":    check_season,
    "heat":      check_heat,
    "beach":     check_beach,
    "trip_type": check_trip_type,
}

```

```python
# MODE 1 - recommend anywhere in egypt with fuzzy matching
def mode1_recommend_egypt(prefs):
    results = []
    tree_log = []
    perfect_matches = []
    close_matches = []

    for dest in destinations:
        passed, checks, tree = backward_chain(dest, prefs)
        tree_log += tree
        tree_log.append("")
        
        fail_count = sum(1 for v in checks.values() if not v)
        
        if passed:
            perfect_matches.append({
                "city": dest,
                "reasons": checks,
                "attractions": attractions.get(dest, []),
                "pack": get_packing_list(dest),
                "fail_count": 0
            })
        else:
            # Track close matches (max 2 failures)
            if fail_count <= 2:
                close_matches.append({
                    "city": dest,
                    "reasons": checks,
                    "attractions": attractions.get(dest, []),
                    "pack": get_packing_list(dest),
                    "fail_count": fail_count,
                    "failed_rules": [k for k,v in checks.items() if not v]
                })
    
    # Sort close matches by fewest failures
    close_matches.sort(key=lambda x: x["fail_count"])
    
    # Return perfect matches if any, otherwise closest matches
    if perfect_matches:
        return perfect_matches, tree_log, "perfect"
    elif close_matches:
        return close_matches, tree_log, f"closest (max {close_matches[0]['fail_count']} failures)"
    else:
        return [], tree_log, "none"
```

```python
# MODE 2 - user picks specific city/cities with fuzzy matching
def mode2_specific_cities(city_list, prefs):
    results = []
    tree_log = []

    for dest in city_list:
        if dest not in destinations:
            results.append({"city": dest, "valid": False, "error": "city not in knowledge base"})
            continue

        passed, checks, tree = backward_chain(dest, prefs)
        tree_log += tree
        tree_log.append("")
        
        fail_count = sum(1 for v in checks.values() if not v)
        failed_rules = [k for k,v in checks.items() if not v]

        entry = {
            "city": dest, 
            "valid": passed, 
            "reasons": checks,
            "fail_count": fail_count,
            "failed_rules": failed_rules
        }
        
        if passed or fail_count <= 2:  # Show if perfect OR close enough
            entry["attractions"] = attractions.get(dest, [])
            entry["pack"] = get_packing_list(dest)
            entry["status"] = "PERFECT MATCH" if passed else f"CLOSE MATCH ({fail_count} failure(s))"
        else:
            entry["status"] = f"NOT RECOMMENDED ({fail_count} failures)"
            
        results.append(entry)

    return results, tree_log
```

```python
# MODE 3 - build a trip itinerary WITH fuzzy matching (max 2 failures)
def mode3_itinerary(prefs, total_days):
    # Get all cities with 0-2 failures (not just perfect matches)
    all_candidates = []
    for dest in destinations:
        passed, checks, _ = backward_chain(dest, prefs)
        fail_count = sum(1 for v in checks.values() if not v)
        if passed or fail_count <= 2:  # FUZZY: perfect OR close enough
            all_candidates.append(dest)
    
    tree_log = build_tree_log(prefs, destinations)

    if not all_candidates:
        return [], "No cities matched your preferences (or close enough) for an itinerary.", tree_log

    # Use candidates for itinerary building
    itinerary = []
    days_used = 0
    used = set()
    matched_sorted = sorted(all_candidates, key=lambda d: days_needed.get(d, 2))

    for city in matched_sorted:
        if city in used:
            continue
        needed = days_needed.get(city, 2)
        if days_used + needed > total_days:
            continue
        itinerary.append(city)
        used.add(city)
        days_used += needed

        for neighbor in compatible_pairs.get(city, []):
            if neighbor in used or neighbor not in all_candidates:
                continue
            n_days = days_needed.get(neighbor, 2)
            if days_used + n_days > total_days:
                continue
            itinerary.append(neighbor)
            used.add(neighbor)
            days_used += n_days

    schedule = []
    day = 1
    for city in itinerary:
        n = days_needed.get(city, 2)
        schedule.append({
            "city": city,
            "day_range": f"Day {day}" if n == 1 else f"Day {day} - {day+n-1}",
            "attractions": attractions.get(city, []),
            "pack": get_packing_list(city),
        })
        day += n

    return schedule, None, tree_log
```

```python
# MODE 4 - what can i do with my budget
# No preference filters, just budget + days.
# Returns all affordable cities ranked cheapest first + combo suggestions.

def mode4_budget_explorer(user_budget, total_days):
    affordable = []
    for dest in destinations:
        dest_budget = get_fact("budget_level", dest)
        if budget_order.get(user_budget, 0) >= budget_order.get(dest_budget, 99):
            affordable.append({
                "city":        dest,
                "budget":      dest_budget,
                "days_needed": days_needed.get(dest, 2),
                "attractions": attractions.get(dest, []),
            })

    affordable.sort(key=lambda x: (budget_order[x["budget"]], x["days_needed"]))

    combos = []
    used   = set()
    affordable_names = [x["city"] for x in affordable]
    for item in affordable:
        city = item["city"]
        if city in used:
            continue
        combo      = [city]
        combo_days = item["days_needed"]
        used.add(city)
        for neighbor in compatible_pairs.get(city, []):
            if neighbor in used or neighbor not in affordable_names:
                continue
            n_days = days_needed.get(neighbor, 2)
            if combo_days + n_days <= total_days:
                combo.append(neighbor)
                combo_days += n_days
                used.add(neighbor)
        combos.append({"cities": combo, "total_days": combo_days})

    return affordable, combos
```

```python
# User Input with NO PREFERENCE option
def get_input(prompt, valid_options, allow_no_pref=True):
    valid = list(valid_options)
    if allow_no_pref:
        valid = valid + ["no preference", "any", "none"]
    while True:
        val = input(prompt).strip().lower()
        if val in valid:
            return val
        print(f"  please enter one of: {', '.join(sorted(valid[:8]))}" + (" + 'no preference'" if allow_no_pref else ""))
```

```python
def main():
    print("\n=== Egypt Travel Assistant ===")
    print("Modes:")
    print("  1 - Recommend me somewhere in Egypt")
    print("  2 - I already know which city/cities I want")
    print("  3 - Build me a trip itinerary")
    print("  4 - What can I do with my budget")

    mode = get_input("Pick a mode (1-4): ", ["1","2","3","4"])

    # MODES 1, 2, 3 - ALL questions have "no preference" option
    budget = get_input("Budget? (low / medium / high / no preference): ", ["low","medium","high","no preference"])
    month  = get_input("Travel month? (january / february / march / april / may / june / july / august / september / october / november / december / no preference): ", 
                       ["january","february","march","april","may","june","july","august","september","october","november","december","no preference"])
    heat   = get_input("Heat tolerance? (low / moderate / high / extreme / no preference): ", ["low","moderate","high","extreme","no preference"])
    beach  = get_input("Want a beach destination? (yes / no / no preference): ", ["yes","no","no preference"])

    all_types = sorted(set(t for types in trip_types.values() for t in types))
    print(f"\nTrip types: {', '.join(all_types)}")
    trip = get_input(f"Preferred trip type? ({', '.join(all_types)} / no preference): ", all_types + ["no preference"])

    # Handle month conversion (None if no preference)
    month_num = None
    if month != "no preference":
        month_num = month_names[month]

    prefs = {
        "budget":         budget,
        "month_num":      month_num,
        "heat_tolerance": heat,
        "wants_beach":    beach,
        "trip_type":      trip,
    }

    # MODE 4
    if mode == "4":
        budget = get_input("Budget? (low / medium / high / no preference): ", ["low","medium","high","no preference"])
        days   = int(input("How many days do you have? ").strip())
        
        # If no preference, set to high to show everything
        if budget == "no preference":
            budget = "high"
            print("\n*** Showing ALL destinations (no budget restriction) ***")
        
        affordable, combos = mode4_budget_explorer(budget, days)
        print(f"\n{len(affordable)} cities within your budget:\n")
        for x in affordable:
            print(f"  {x['city'].replace('_',' ').title()} ({x['budget']} budget, {x['days_needed']} days)")
        print("\nSuggested combinations that fit your days:")
        for c in combos:
            print(f"  {' + '.join(c['cities'])} = {c['total_days']} days")
        return
    
    # MODE 1
    if mode == "1":
        results, tree_log, match_type = mode1_recommend_egypt(prefs)
        print("\n--- Reasoning (search tree) ---\n")
        for line in tree_log:
            print(line)
        print("\n--- Results ---\n")
        
        if not results:
            print("No destinations matched within 2 failed criteria.")
            print("Try adjusting your preferences.")
            return
        
        if match_type == "perfect":
            print("✅ PERFECT MATCHES (0 failures):\n")
        else:
            print(f"⚠️ CLOSEST MATCHES ({match_type}):\n")
            
        for r in results:
            if r.get("fail_count", 0) > 0:
                print(f"  {r['city'].replace('_',' ').title()} - {r['fail_count']} issue(s): {', '.join(r.get('failed_rules', []))}")
            else:
                print(f"  {r['city'].replace('_',' ').title()} - PERFECT!")
            print("  Attractions:")
            for a in r["attractions"][:5]:  # Show top 5 attractions
                print(f"    - {a}")
            print("  Pack:")
            for p in r["pack"]:
                print(f"    - {p}")
            print()

       # MODE 2
    elif mode == "2":
        raw = input("Enter city names separated by commas: ").strip().lower()
        cities = [c.strip().replace(" ", "_") for c in raw.split(",")]
        results, tree_log = mode2_specific_cities(cities, prefs)
        print("\n--- Reasoning (search tree) ---\n")
        for line in tree_log:
            print(line)
        print("\n--- Results ---\n")
        for r in results:
            name = r["city"].replace("_", " ").title()
            print(f"  {name}: {r['status']}")
            if r.get("fail_count", 0) > 0 and r.get("failed_rules"):
                print(f"    Issues: {', '.join(r['failed_rules'])}")
            if r.get("attractions"):
                print("    Top attractions:")
                for a in r["attractions"][:3]:
                    print(f"      - {a}")
                print("    Pack:")
                for p in r.get("pack", [])[:4]:
                    print(f"      - {p}")
            print()

    # MODE 3
    elif mode == "3":
        days = int(input("How many days do you have? ").strip())
        schedule, err, tree_log = mode3_itinerary(prefs, days)
        print("\n--- Reasoning (search tree) ---\n")
        for line in tree_log:
            print(line)
        print("\n--- Itinerary ---\n")
        if err:
            print(err)
            return
        for stop in schedule:
            print(f"  {stop['day_range']}: {stop['city'].replace('_',' ').title()}")
            for a in stop["attractions"]: print(f"    - {a}")
            print()
```
